```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t=0.2           #for NACA airfoil 0012 symmetric airfoil
x=np.linspace(0,1,1000)

yt= 5*t*(0.2969*np.sqrt(x) - 0.1260*x - 0.3516*x**2 + 0.2843*x**3 - 0.1015*x**4)

plt.plot(x,yt , label="upper surface", color="black")
plt.plot(x,-yt , label="lower surface", color="red")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title("NACA 0012 Airfoil")
plt.axis("equal")  #to make the 1 unit at x axis equal to 1 unit of y axis
plt.legend()
plt.grid(True)
plt.show()
```

# Now for unsymmetric Airfoil

```python
import numpy as np
import matplotlib.pyplot as plt

# NACA 2412
m=0.02  #max camber
p=0.4  # location of max camber
t=0.12  # thickness

x=np.linspace(0,1,200)
yt= 5*t*(0.2969*np.sqrt(x) - 0.1260*x - 0.3516*x**2 + 0.2843*x**3 - 0.1015*x**4)  #thickness distribution
yc=np.where( x<p, 
             m/p**2 * (2*p*x - x**2),
             m/(1-p)**2 * ((1-2*p) + 2*p*x - x**2))

xu=x
yu=yc + yt
xl=x
yl=yc - yt
plt.plot(xu,yu, label="upper surface")
plt.plot(xl,yl, label="lower surface")
plt.plot(x,yt ,'--', label="NACA 0012", color="black")
plt.plot(x,-yt ,'--', color="black")
plt.plot(x,yc,'--', label="camber line", color="grey")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title("NACA 2412 Airfoil")
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.show()
```

# 5 digit Airfoil

```python
import numpy as np
import matplotlib.pyplot as plt

#NACA 23012
m=0.2025
k1=15.957
t=0.12

x=np.linspace(0,1,200)

yt=5*t*(0.2969*np.sqrt(x) - 0.1260*x - 0.3516*x**2 + 0.2843*x**3 - 0.1015*x**4)
yc=np.where(x<m,
            (k1/6)*(x**3 - 3*m*x**2 + m**2*(3-m)*x),
            (k1*m**3/6)*(1-x))

xu=x
yu=yc + yt
xl=x
yl=yc - yt

plt.plot(xu,yu, label="upper surface")
plt.plot(xl,yl, label="lower surface")
plt.plot(x,yc,'--', label="camber line", color="grey")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title("NACA 23012 Airfoil")
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def naca_4digit(m,p,t,n_points=200):
    x=np.linspace(0,1,n_points)

    yt= 5*t*(0.2969*np.sqrt(x) - 0.1260*x - 0.3516*x**2 + 0.2843*x**3 - 0.1015*x**4)  #thickness distribution
    yc=np.where( x<p, 
             m/p**2 * (2*p*x - x**2),
             m/(1-p)**2 * ((1-2*p) + 2*p*x - x**2))
    yu=yc + yt
    yl=yc - yt
    return x,yu,yl,yc
```

```python
x,yu,yl,yc=naca_4digit(m=0.02,p=0.4,t=0.12)
plt.plot(xu,yu, label="upper surface")
plt.plot(xl,yl, label="lower surface")
plt.plot(x,yc,'--', label="camber line", color="grey")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title("NACA 2412 Airfoil")
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.show()
```

```python
x,yu,yl,yc=naca_4digit(m=0.04,p=0.4,t=0.12)
plt.plot(xu,yu, label="upper surface")
plt.plot(xl,yl, label="lower surface")
plt.plot(x,yc,'--', label="camber line", color="grey")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title("NACA 4412 Airfoil")
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.show()
```
