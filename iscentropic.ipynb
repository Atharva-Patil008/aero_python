```python
import numpy as np
import matplotlib.pyplot as plt

gamma = 1.4
M = np.linspace(0,5,100)
T0_over_T = 1 + (gamma -1)/2 * M**2

plt.plot(M, T0_over_T)
plt.xlabel("Mach Number(M)")
plt.ylabel("T0/T")
plt.title("Stagnation to static temperature ratio")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

gamma = 1.4
M=np.linspace(0,5,100)

T0_over_T = 1 + (gamma -1)/2 * M**2
P0_over_P= (T0_over_T)**(gamma/(gamma-1))
rho0_over_rho= (T0_over_T)**(1/(gamma-1))

plt.plot(M, T0_over_T, label="T0/T")
plt.plot(M, P0_over_P, label="P0/P")
plt.plot(M, rho0_over_rho, label="rho0/rho")
plt.yscale("log")
plt.xlabel("Mach No.(M)")
plt.ylabel("ratio")
plt.title("iscentropic relations")
plt.legend()
plt.grid(True)
plt.show()
```
